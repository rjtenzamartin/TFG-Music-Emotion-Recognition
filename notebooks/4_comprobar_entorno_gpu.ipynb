# **Comprobar entorno GPU**

```python
import sys
import importlib
import torch

print(f"Python versión: {sys.version.split()[0]}")
print("-" * 50)

# 1. Comprobar GPU
print("CHEQUEO DE ACELERACIÓN POR HARDWARE:")
cuda_ok = torch.cuda.is_available()
print(f"   ¿PyTorch detecta CUDA?: {'SÍ' if cuda_ok else 'NO'}")

if cuda_ok:
    print(f"   GPU detectada: {torch.cuda.get_device_name(0)}")
    print(f"   VRAM Total: {torch.cuda.get_device_properties(0).total_memory / 1e9:.2f} GB")
    print(f"   Versión de CUDA: {torch.version.cuda}")
else:
    print("   ATENCIÓN: No se detecta GPU. El entrenamiento será muy lento.")

print("-" * 50)

# 2. Inventario de librerías críticas
librerias = {
    "torch": "Base de IA",
    "torchaudio": "Procesado de audio en GPU",
    "librosa": "Análisis de audio",
    "pandas": "Gestión de TSV",
    "numpy": "Cálculo numérico",
    "sklearn": "Métricas de evaluación",
    "matplotlib": "Gráficos",
    "tqdm": "Barras de progreso"
}

print("ESTADO DE LIBRERÍAS:")
faltantes = []

for lib, desc in librerias.items():
    try:
        mod = importlib.import_module(lib)
        ver = getattr(mod, "__version__", "N/A")
        print(f"   {lib:<12} (v{ver:<8}) - {desc}")
    except ImportError:
        print(f"   {lib:<12} (FALTA)    - {desc}")
        faltantes.append(lib)

print("-" * 50)
if not faltantes:
    print("¡TODO LISTO! Tu entorno está preparado para la Fase 2.")
else:
    print(f"ACCIÓN REQUERIDA: Instala las librerías que faltan usando:")
    print(f"   !pip install {' '.join(faltantes)}")
```
